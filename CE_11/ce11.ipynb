```python
def normalize_longitude(lon: float) ->float:
    """ Normalize a longitude value to be in a range of (-180, 180)."""
    if lon > 180:
        return lon - 360
    elif lon < -180:
        return lon + 360
    else:
        return lon  
```

```python
def test_normalized_longitude():
    expected = {
        360: 0, 
        -190: 170,
    }
    actual= dict()
    
    for val in expected:
        actual_longitude= normalize_longitude(val)
        actual[val]= actual_longitude
        
        
    
    assert actual == expected
    
        
    
```

```python
test_normalized_longitude()
```
