```python
class house: 
    def __init__(self,name, lat, lon):
        self.name = name
        self.lat = lat
        self.lon = lon

    def summary(self):
        return f"{self.name}: ({self.lat:.4f}, {self.lon: .4f})"


```

```python
pt1 = house('Site A ', 35.04, -86.32)
```
