```python
import shapely
```

```python
class person:
    """Individual with a name and contact information"""
    def __init__(self, name, contact):
        self.name = name
        self.contact = contact 
    def __repr__(self):
        return f"person(name= '{self.name}', phone= '{self.contact}')"

class Home:
    """Home address with latitude and longitude"""
    def __init__(self, address, lat, lon):
        self.address = address
        self.coords = shapely.Point(lat, lon)
        self.residents = []

    def add_resident(self, person): 
        self.residents.append(person)
        
    def get_contact_information(self): 
        return [resident.contact for resident in self.residents]
```

```python
p1 = person("Alice", "304-555-0101")
p2 = person("Bob", "703-451-123")
my_home= Home("123 Sesame Court", 40.7128, -74.0060)
my_home.add_resident(p1)
my_home.add_resident(p2)
print(f"Contacts for {my_home.address}: {my_home.get_contact_information()}")
```
